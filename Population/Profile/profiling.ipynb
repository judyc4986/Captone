```python
# ----------------------------------------------------------
# Washington EV Infrastructure Capstone – Automated Profiling
# Save Reports Directly to Desktop
# ----------------------------------------------------------

import os
import pandas as pd
from ydata_profiling import ProfileReport

# --- Change working directory to Desktop ---
os.chdir("/Users/judycheng/Desktop")
print("📂 Reports will be saved to:", os.getcwd())

# --- 1️⃣ Supercharger dataset ---
supercharger_path = "/Users/judycheng/Desktop/supercharger in washington state.xls"
supercharger_df = pd.read_excel(supercharger_path)

supercharger_profile = ProfileReport(
    supercharger_df,
    title="🏁 Tesla Supercharger Network Profile (Washington)",
    explorative=True
)
supercharger_profile.to_file("Report_Supercharger_Profile.html")
print("✅ Supercharger profile saved to Desktop")


# --- 2️⃣ Population dataset ---
population_path = "/Users/judycheng/Desktop/Population 2024 age 25 to 59.xlsx"
population_df = pd.read_excel(population_path)

population_profile = ProfileReport(
    population_df,
    title="👥 Washington Population Profile (Age 25–59)",
    explorative=True
)
population_profile.to_file("Report_Population_Profile.html")
print("✅ Population profile saved to Desktop")


# --- 3️⃣ EV VIN registration dataset ---
ev_path = "/Users/judycheng/Desktop/coordinates_output.xlsm"
ev_df = pd.read_excel(ev_path, sheet_name=0)  # adjust sheet_name if needed

ev_profile = ProfileReport(
    ev_df,
    title="🔋 EV VIN Registration Profile (Battery Electric Vehicles)",
    explorative=True
)
ev_profile.to_file("Report_EV_VIN_Profile.html")
print("✅ EV VIN registration profile saved to Desktop")
```
